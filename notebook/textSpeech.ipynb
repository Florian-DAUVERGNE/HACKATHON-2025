```python
from gtts import gTTS
from pydub import AudioSegment
from pydub.playback import play

# Texte à convertir en parole
text = """Période : En semaine.Date : Vendredi 27 Décembre 2024.​Vendredi 27 Décembre, 
aucun train ne circule entre Paris gare de Lyon surface et Montargis/ Montereau à partir de 22h50.Derniers trains au départ de Paris gare de Lyon surface en direction de Montereau et Montargis:
- GAMO départ Paris gare de Lyon Surface à 22h16, arrivée Montargis à 23h55.
- KUMO départ Paris gare de Lyon Surface à 22h46, arrivée Montereau à 23h44.Derniers trains au départ de Montereau et Montargis en direction de Paris gare de Lyon Surface:
- POMA départ Montargis à 21h01, arrivée à Paris gare de Lyon Surface à 22h43.- PUMA départ Montereau à 22h14, arrivée à Paris gare de Lyon Surface à 23h13.Les horaires du calculateur d'itinéraire tiennent compte des travaux.Motif: travaux"""

# Langue (par exemple, 'fr' pour le français)
language = 'fr'

# Conversion du texte en audio
speech = gTTS(text=text, lang=language, slow=False)
speech.save("output.mp3")

# Charger le fichier audio avec pydub
audio = AudioSegment.from_file("output.mp3")

# Modifier la vitesse (par exemple, 1.5x pour accélérer ou 0.75x pour ralentir)
speed = 1.5  # Remplacez par la vitesse souhaitée
audio = audio._spawn(audio.raw_data, overrides={
    "frame_rate": int(audio.frame_rate * speed)
}).set_frame_rate(audio.frame_rate)

# Sauvegarder le fichier audio modifié
audio.export("output_modified.mp3", format="mp3")

# Lecture du fichier audio modifié (facultatif)
play(audio)
```
