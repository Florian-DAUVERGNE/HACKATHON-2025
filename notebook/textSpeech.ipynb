```python
from gtts import gTTS
import os

# Texte à convertir en parole
text = "Période : En semaine.Date : Vendredi 27 Décembre 2024.​Vendredi 27 Décembre, aucun train ne circule entre Paris gare de Lyon surface et Montargis/ Montereau à partir de 22h50.Derniers trains au départ de Paris gare de Lyon surface en direction de Montereau et Montargis:- GAMO départ Paris gare de Lyon Surface à 22h16, arrivée Montargis à 23h55.- KUMO départ Paris gare de Lyon Surface à 22h46, arrivée Montereau à 23h44.Derniers trains au départ de Montereau et Montargis en direction de Paris gare de Lyon Surface:- POMA départ Montargis à 21h01, arrivée à Paris gare de Lyon Surface à 22h43.- PUMA départ Montereau à 22h14, arrivée à Paris gare de Lyon Surface à 23h13.Les horaires du calculateur d'itinéraire tiennent compte des travaux.Motif: travaux"

# Langue (par exemple, 'fr' pour le français)
language = 'fr'

# Conversion du texte en audio
speech = gTTS(text=text, lang=language, slow=False)

# Sauvegarde du fichier audio
speech.save("output.mp3")

# Lecture du fichier audio (facultatif)
os.system("start output.mp3")
```

```python
from google.cloud import texttospeech
import os

# Définir le chemin vers le fichier de clé JSON
os.environ["GOOGLE_APPLICATION_CREDENTIALS"] = "path/to/your/service-account-key.json"

# Initialiser le client Text-to-Speech
client = texttospeech.TextToSpeechClient()

# Texte à convertir en parole
text = """Période : En semaine. Date : Vendredi 27 Décembre 2024.
Vendredi 27 Décembre, aucun train ne circule entre Paris gare de Lyon surface et Montargis/Montereau à partir de 22h50.
Derniers trains au départ de Paris gare de Lyon surface en direction de Montereau et Montargis :
- GAMO départ Paris gare de Lyon Surface à 22h16, arrivée Montargis à 23h55.
- KUMO départ Paris gare de Lyon Surface à 22h46, arrivée Montereau à 23h44.
Derniers trains au départ de Montereau et Montargis en direction de Paris gare de Lyon Surface :
- POMA départ Montargis à 21h01, arrivée à Paris gare de Lyon Surface à 22h43.
- PUMA départ Montereau à 22h14, arrivée à Paris gare de Lyon Surface à 23h13.
Les horaires du calculateur d'itinéraire tiennent compte des travaux. Motif : travaux."""

# Configurer la requête
synthesis_input = texttospeech.SynthesisInput(text=text)

# Configurer la voix (langue et type de voix)
voice = texttospeech.VoiceSelectionParams(
    language_code="fr-FR",  # Français
    name="fr-FR-Neural2-A",  # Voix neuronale
    ssml_gender=texttospeech.SsmlVoiceGender.FEMALE,
)

# Configurer le format audio
audio_config = texttospeech.AudioConfig(
    audio_encoding=texttospeech.AudioEncoding.MP3
)

# Envoyer la requête à l'API
response = client.synthesize_speech(
    input=synthesis_input, voice=voice, audio_config=audio_config
)

# Sauvegarder le fichier audio
with open("output.mp3", "wb") as out:
    out.write(response.audio_content)
    print("Le fichier audio a été sauvegardé sous le nom 'output.mp3'.")

# Lecture du fichier audio (facultatif)
import os
os.system("start output.mp3")
```

```python
from TTS.api import TTS

# Charger un modèle pré-entraîné
tts = TTS(model_name="tts_models/fr/css10/vits", progress_bar=True, gpu=False)

# Texte à convertir en parole
text = """Période : En semaine. Date : Vendredi 27 Décembre 2024.
Vendredi 27 Décembre, aucun train ne circule entre Paris gare de Lyon surface et Montargis/Montereau à partir de 22h50.
Derniers trains au départ de Paris gare de Lyon surface en direction de Montereau et Montargis :
- GAMO départ Paris gare de Lyon Surface à 22h16, arrivée Montargis à 23h55.
- KUMO départ Paris gare de Lyon Surface à 22h46, arrivée Montereau à 23h44.
Derniers trains au départ de Montereau et Montargis en direction de Paris gare de Lyon Surface :
- POMA départ Montargis à 21h01, arrivée à Paris gare de Lyon Surface à 22h43.
- PUMA départ Montereau à 22h14, arrivée à Paris gare de Lyon Surface à 23h13.
Les horaires du calculateur d'itinéraire tiennent compte des travaux. Motif : travaux."""

# Convertir le texte en audio et sauvegarder dans un fichier
tts.tts_to_file(text=text, file_path="output.wav")

# Lecture du fichier audio (facultatif)
import os
os.system("start output.wav")
```
